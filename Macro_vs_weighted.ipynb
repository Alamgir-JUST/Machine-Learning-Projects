## **Macro vs weighted**

### **Create a Sample Dataset**

```python
# Required Libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score
from sklearn.datasets import make_classification

# Step 1: Create a Sample Dataset
X, y = make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=42)

```

```python
X
```

```python
y
```

```python
# Step 2: Split the Dataset into Training and Testing Sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Step 3: Initialize and Train a Random Forest Classifier
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train, y_train)
```

```python
# Step 4: Predict on the Test Set
y_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import classification_report

# Step 5: Print the Classification Report with 5 decimal places
report = classification_report(y_test, y_pred, digits=5)

# Print the formatted report
print(report)
```

```python
# Step 5: Calculate F1 Scores (Macro and Weighted)
f1_macro = f1_score(y_test, y_pred, average='macro')
f1_weighted = f1_score(y_test, y_pred, average='weighted')

# Output the Results
print(f"F1 Score (Macro): {f1_macro:.4f}")
print(f"F1 Score (Weighted): {f1_weighted:.4f}")
```

```python
from sklearn.metrics import accuracy_score

# Calculate Macro Accuracy
accuracy_macro = accuracy_score(y_test, y_pred)

# Calculate Weighted Accuracy
accuracy_weighted = accuracy_score(y_test, y_pred, normalize=True)  # Default normalize=True for weighted accuracy

# Print the results
print(f"Macro Accuracy: {accuracy_macro:.5f}")
print(f"Weighted Accuracy: {accuracy_weighted:.5f}")
```

```python
from sklearn.metrics import precision_score

# Calculate Macro Precision (average precision for all classes)
precision_macro = precision_score(y_test, y_pred, average='macro')

# Calculate Weighted Precision (average precision weighted by class support)
precision_weighted = precision_score(y_test, y_pred, average='weighted')

# Print the results
print(f"Macro Precision: {precision_macro:.5f}")
print(f"Weighted Precision: {precision_weighted:.5f}")
```

```python
from sklearn.metrics import recall_score


# Calculate Macro Recall (average recall for all classes)
recall_macro = recall_score(y_test, y_pred, average='macro')

# Calculate Weighted Recall (average recall weighted by class support)
recall_weighted = recall_score(y_test, y_pred, average='weighted')

# Print the results
print(f"Macro Recall: {recall_macro:.5f}")
print(f"Weighted Recall: {recall_weighted:.5f}")
```

## **Confusion Matrix**

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Calculate the Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# Plot the Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()

# Print the confusion matrix
print("Confusion Matrix:")
print(cm)
```
